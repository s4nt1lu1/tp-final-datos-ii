## Carga del dataset

```python
import sys
from pathlib import Path
import pandas as pd

# Anclar la raíz del proyecto para importar rutas desde src/
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))

from src.auxiliar_functions import PROCESSED_DATA_DIR

ruta_dataset = PROCESSED_DATA_DIR / "dataset_unificado_limpio.csv"
df = pd.read_csv(ruta_dataset, encoding="utf-8", low_memory=False)

print(f"Archivo cargado: {ruta_dataset}")
print(f"Dimensiones (filas, columnas): {df.shape}")
```

## Columnas del dataset

```python
print(f"Cantidad total de columnas: {len(df.columns)}\n")
for i, col in enumerate(df.columns, start=1):
    print(f"{i:>3}. {col}")
```

## Eliminación de variables "Meta"

```python
cols_meta = [col for col in df.columns if "meta" in col.lower()]

print(f"Columnas que contienen 'Meta' a eliminar ({len(cols_meta)}):")
for col in cols_meta:
    print(f"  - {col}")

df = df.drop(columns=cols_meta)

print(f"\nNuevas dimensiones (filas, columnas): {df.shape}")
```
